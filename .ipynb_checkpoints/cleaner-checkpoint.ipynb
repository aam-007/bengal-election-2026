```python
import camelot
import pandas as pd

PDF_PATH = r"C:\Users\Admin\mainframe\Programming\bengal_election_prediction_model\download\1951.pdf"

print("STARTING TABLE EXTRACTION...", flush=True)

tables = camelot.read_pdf(
    PDF_PATH,
    pages="all",
    flavor="stream"   # ECI tables usually need stream
)

print(f"Tables found: {tables.n}", flush=True)

if tables.n == 0:
    print("❌ NO TABLES FOUND — STOP")
    exit()

all_tables = []

for i, table in enumerate(tables):
    df = table.df
    df["source_table"] = i
    all_tables.append(df)

final_df = pd.concat(all_tables, ignore_index=True)
final_df.to_csv("raw_tables.csv", index=False)

print("RAW TABLES SAVED AS raw_tables.csv")
```
